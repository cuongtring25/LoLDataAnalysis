```python
import os
from dotenv import load_dotenv
import requests
```

```python
load_dotenv()
API_KEY = os.getenv("API_KEY")
game_name = "Duro"
game_tag = "Gen"
print(API_KEY)
```

```python
url = f'https://asia.api.riotgames.com/riot/account/v1/accounts/by-riot-id/{game_name}/{game_tag}?api_key={API_KEY}'

request = requests.get(url=url)
data = request.json()

print(data)
```

```python
puuid = data['puuid']
print(puuid)
```

```python
matchs_url = f'https://asia.api.riotgames.com/lol/match/v5/matches/by-puuid/{puuid}/ids?start=0&count=5&api_key={API_KEY}'

match_ids = requests.get(url=matchs_url).json()
selected_match = match_ids[1]
print(match_ids[1])
```

```python
match_url  = f'https://asia.api.riotgames.com/lol/match/v5/matches/{selected_match}?api_key={API_KEY}'
raw_match_data = requests.get(url=match_url).json()
print(raw_match_data)
```

```python
attribute_in_game ={}
for p in raw_match_data["info"]["participants"]:
        if p["puuid"] == puuid:
            attribute_in_game = {
                "kills": p["kills"],
                "deaths": p["deaths"],
                "assists": p["assists"],
                "cs": p["totalMinionsKilled"],
                "gold": p["goldEarned"],
                "damage": p["totalDamageDealtToChampions"],
                "time": raw_match_data["info"]["gameDuration"] / 60
            }
print(attribute_in_game)
```
